```python
import pandas as pd, glob
for f in sorted(glob.glob("results/results_*.csv")):
    df = pd.read_csv(f)
    print(f"\n{f.split('/')[-1]}")
    print(f"  rows:           {len(df)}")
    print(f"  parse_ok:       {df['parse_ok'].sum()} / {len(df)}")
    print(f"  empty answers:  {(df['chosen_position'].fillna('') == '').sum()}")
    print(f"  mean latency:   {df['latency_s'].mean():.2f}s")
```

```python
import glob, subprocess
for f in sorted(glob.glob("results/results_*.csv")):
    base = f.split("/")[-1].replace("results_", "analysis_").replace(".csv", ".txt")
    out = f"/content/drive/MyDrive/caste_audit_results/{base}"
    with open(out, "w") as fh:
        result = subprocess.run(
            ["python3", "src/03b_analyze.py", "--file", f],
            capture_output=True, text=True
        )
        fh.write(result.stdout)
        fh.write("\n--- STDERR ---\n")
        fh.write(result.stderr)
    print(f"saved: {out}")

print("\n=== Final state of Drive folder ===")
!ls -la /content/drive/MyDrive/caste_audit_results/
```
